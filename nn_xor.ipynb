```python
import numpy as np

# ==========================================
# 1. THE DATASET (Training Data)
# ==========================================
# X = The input combinations (4 samples, 2 features each)
X_train = np.array([[0, 0],[0,1], [1,0],[1,1]])

# y = The expected correct answers (4 samples, 1 output each)
y_train = np.array([[0],[1],[1],[0]])


```

```python
# ==========================================
# 2. ACTIVATION FUNCTION (The Neuron's Switch)
# ==========================================
# Sigmoid squashes any number down to a value between 0 and 1 (like a probability)
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# The derivative (slope) of sigmoid, used during Backpropagation to find errors
def sigmoid_derivative(x):
    return x * (1 - x)
```

```python
# ==========================================
# 3. INITIALISING THE BRAIN (Weights & Biases)
# ==========================================
# Setting a random seed so the results are exactly reproducible
np.random.seed(42)

# Neural Network Architecture: 2 Inputs -> 3 Hidden Neurons -> 1 Output
# Weights are random starting points; biases act like adjustments to trigger neurons
weights_hidden = np.random.uniform(size=(2, 3))
bias_hidden = np.zeros((1, 3))

weights_output = np.random.uniform(size=(3, 1))
bias_output = np.zeros((1, 1))

learning_rate = 0.5
```

```python
# ==========================================
# 4. THE TRAINING LOOP
# ==========================================
print("--- Training Started ---")

for epoch in range(10000): # Run the "Guess and Correct" loop 10,000 times

    # ---- A. FORWARD PROPAGATION (The Guess) ----
    # Step 1: Push data from input layer into hidden layer
    hidden_activation = np.dot(X_train, weights_hidden) + bias_hidden
    hidden_output = sigmoid(hidden_activation)

    # Step 2: Push hidden layer outputs into the final output layer
    output_activation = np.dot(hidden_output, weights_output) + bias_output
    predicted_output = sigmoid(output_activation)

    # ---- B. CALCULATE ERROR ----
    # How far off was the guess compared to the real y values?
    error = y_train - predicted_output

    # Print out our progress every 2000 loops
    if epoch % 2000 == 0:
        average_loss = np.mean(np.abs(error))
        print(f"Epoch {epoch:5d} | Average Error Size: {average_loss:.5f}")

    # ---- C. BACKPROPAGATION (The Correction) ----
    # Step 1: Calculate error direction and size at the output layer
    d_output = error * sigmoid_derivative(predicted_output)

    # Step 2: Pass that error backward to see how much the hidden layer is to blame
    error_hidden_layer = np.dot(d_output, weights_output.T) # Notice the matrix transposition (.T)!
    d_hidden = error_hidden_layer * sigmoid_derivative(hidden_output)

    # ---- D. UPDATING THE KNOBS (Gradient Descent) ----
    # Adjust weights and biases using the learning rate
    weights_output += np.dot(hidden_output.T, d_output) * learning_rate
    bias_output += np.sum(d_output, axis=0, keepdims=True) * learning_rate

    weights_hidden += np.dot(X_train.T, d_hidden) * learning_rate
    bias_hidden += np.sum(d_hidden, axis=0, keepdims=True) * learning_rate

print("\n--- Training Complete ---\n")
```

```python
# ==========================================
# 5. TESTING THE MODEL (Final Testing Evaluation)
# ==========================================
print("Final Predictions vs Reality:")
for i in range(len(X_train)):
    print(f"Inputs: {X_train[i]} | Targeted: {y_train[i][0]} | Network Guess: {predicted_output[i][0]:.4f}")
```
